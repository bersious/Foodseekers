```python
# 2. TẢI DỮ LIỆU TỰ ĐỘNG TỪ GITHUB REPO CỦA ĐỘI (KHÔNG CẦN UPLOAD THỦ CÔNG!)
import os
import pandas as pd

if not os.path.exists('Foodseekers'):
    !git clone https://github.com/bersious/Foodseekers.git

train_df = pd.read_csv('Foodseekers/data/train.csv')
test_df = pd.read_csv('Foodseekers/data/test.csv')

print(f"[+] Đã tải thành công dữ liệu:")
print(f"  - Train: {train_df.shape[0]} dòng")
print(f"  - Test: {test_df.shape[0]} dòng")
print(train_df['label'].value_counts())
```

```python
# 3. CHUẨN BỊ MÔ HÌNH VÀ DATA LOADER CÓ CÂN BẰNG LỚP (CLASS WEIGHTS)
import numpy as np
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from transformers import AutoTokenizer, AutoModelForSequenceClassification, get_linear_schedule_with_warmup
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, classification_report
from tqdm.auto import tqdm

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
MODEL_NAME = 'FacebookAI/xlm-roberta-base'
BATCH_SIZE = 16
MAX_LEN = 256
EPOCHS = 3
LR = 2e-5

CLASSES = ['BENIGN', 'JAILBREAK', 'PROMPT_INJECTION']
LABEL2ID = {c: i for i, c in enumerate(CLASSES)}
ID2LABEL = {i: c for i, c in enumerate(CLASSES)}
train_df['target'] = train_df['label'].map(LABEL2ID)

# Phạt nặng lỗi ở nhãn hiếm PROMPT_INJECTION (7%) để tối đa hóa Macro F1
class_counts = train_df['label'].value_counts()[CLASSES].values
class_weights = len(train_df) / (len(CLASSES) * class_counts)
weights_tensor = torch.tensor(class_weights, dtype=torch.float).to(device)
print("Trọng số cân bằng lớp (Class weights):", weights_tensor)

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

class PromptDataset(Dataset):
    def __init__(self, texts, labels=None, tokenizer=None, max_len=256):
        self.texts = texts
        self.labels = labels
        self.tokenizer = tokenizer
        self.max_len = max_len
    def __len__(self):
        return len(self.texts)
    def __getitem__(self, idx):
        text = str(self.texts[idx])
        enc = self.tokenizer(text, truncation=True, max_length=self.max_len, padding='max_length', return_tensors='pt')
        item = {'input_ids': enc['input_ids'].squeeze(0), 'attention_mask': enc['attention_mask'].squeeze(0)}
        if self.labels is not None:
            item['label'] = torch.tensor(self.labels[idx], dtype=torch.long)
        return item

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
train_idx, val_idx = next(skf.split(train_df, train_df['target']))
df_tr = train_df.iloc[train_idx].reset_index(drop=True)
df_va = train_df.iloc[val_idx].reset_index(drop=True)

train_loader = DataLoader(PromptDataset(df_tr['text'].values, df_tr['target'].values, tokenizer, MAX_LEN), batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(PromptDataset(df_va['text'].values, df_va['target'].values, tokenizer, MAX_LEN), batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(PromptDataset(test_df['text'].values, None, tokenizer, MAX_LEN), batch_size=BATCH_SIZE, shuffle=False)
```

```python
# 4. HUẤN LUYỆN MÔ HÌNH VÀ LƯU CHECKPOINT TỐT NHẤT
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=3, id2label=ID2LABEL, label2id=LABEL2ID).to(device)
criterion = nn.CrossEntropyLoss(weight=weights_tensor)
optimizer = AdamW(model.parameters(), lr=LR, weight_decay=0.01)
total_steps = len(train_loader) * EPOCHS
scheduler = get_linear_schedule_with_warmup(optimizer, num_warmup_steps=int(total_steps * 0.1), num_training_steps=total_steps)

best_macro_f1 = 0.0
for epoch in range(EPOCHS):
    model.train()
    for batch in tqdm(train_loader, desc=f'Epoch {epoch+1}/{EPOCHS} [Train]'):
        optimizer.zero_grad()
        out = model(input_ids=batch['input_ids'].to(device), attention_mask=batch['attention_mask'].to(device))
        loss = criterion(out.logits, batch['label'].to(device))
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        scheduler.step()
        
    model.eval()
    preds, trues = [], []
    with torch.no_grad():
        for batch in tqdm(val_loader, desc=f'Epoch {epoch+1}/{EPOCHS} [Val]'):
            out = model(input_ids=batch['input_ids'].to(device), attention_mask=batch['attention_mask'].to(device))
            p = torch.argmax(out.logits, dim=1).cpu().numpy()
            preds.extend(p)
            trues.extend(batch['label'].numpy())
            
    score = f1_score(trues, preds, average='macro')
    print(f'\n[+] Kết thúc Epoch {epoch+1} — Validation Macro F1: {score:.4f}')
    if score > best_macro_f1:
        best_macro_f1 = score
        torch.save(model.state_dict(), 'best_model.pt')
        print(f'[*] Đã lưu mô hình kỷ lục với Macro F1: {best_macro_f1:.4f}!')

print(f'\n🏆 ĐIỂM CAO NHẤT ĐẠT ĐƯỢC: {best_macro_f1:.4f}')
```

```python
# 5. DỰ ĐOÁN TẬP TEST VÀ TỰ ĐỘNG TẢI FILE SUBMISSION VỀ MÁY
from google.colab import files

print("[*] Đang tải checkpoint tốt nhất để dự đoán 6.000 dòng test...")
model.load_state_dict(torch.load('best_model.pt'))
model.eval()

test_preds = []
with torch.no_grad():
    for batch in tqdm(test_loader, desc='Predicting Test'):
        out = model(input_ids=batch['input_ids'].to(device), attention_mask=batch['attention_mask'].to(device))
        p = torch.argmax(out.logits, dim=1).cpu().numpy()
        test_preds.extend(p)

submission = pd.DataFrame({
    'id': test_df['id'],
    'label': [ID2LABEL[i] for i in test_preds]
})
submission.to_csv('submission_colab_xlm_roberta.csv', index=False)
print("[✅] Đã tạo xong file: submission_colab_xlm_roberta.csv (đúng 6.000 dòng)!")
print(submission['label'].value_counts())

# Tự động bật cửa sổ tải file về máy tính cho Thiện Nhân
files.download('submission_colab_xlm_roberta.csv')
```
